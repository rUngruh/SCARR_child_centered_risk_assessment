```python
import sys
import pandas as pd
import os
import time

# abspath = os.path.abspath(__file__)
# dname = os.path.dirname(abspath)
# os.chdir(dname)


from pathlib import Path
dataset_dir = Path('../data')

import utils.genre_processing as gp


save_dir = dataset_dir / 'processed' / 'MAL'
data_dir = dataset_dir / 'raw' / 'MyAnimeList'

cover_dir = save_dir / 'cover_art'

item_raw_path = data_dir / 'anime_cleaned.csv'

if not os.path.exists(save_dir):
    os.makedirs(save_dir)

# import argparse
# parser = argparse.ArgumentParser(description='Process MAL dataset.')
# parser.add_argument('--experiment_year', type=int, default=2015, help='Year to filter interactions for the experiment.')
# args = parser.parse_args()

# experiment_year = args.experiment_year

experiment_year = 2015

itemPath = save_dir / f'items_{experiment_year}_enriched.tsv'

items = pd.read_csv(itemPath, delimiter='\t') # Index(['item_id', 'title', 'title_english', 'age_rating_raw', 'genres'], dtype='str')

userPath = save_dir / f'users_{experiment_year}.tsv'

users = pd.read_csv(userPath, delimiter='\t') # Index(['user_id', 'gender', 'birth_age', 'age'], dtype='str')
```

```python
items['genres'] = items['genres'].apply(lambda x: x.split(',') if isinstance(x, str) else [])
```

```python

genre_counts = items['genres'].explode().value_counts()
genre_counts
```

```python
missing_covers  = 0
missing_synopsis = 0
for index, row in items.iterrows():
    item_id = row['item_id']
    cover_path = cover_dir / f'{item_id}.jpg'
    
    # Check how many covers are missing
    if not os.path.exists(cover_path):
        missing_covers += 1
    
    description = row.get('synopsis', None)
    # Check how many descriptions are missing
    if pd.isna(description) or description.strip() == '':
        missing_synopsis += 1
        
print(f"Total items: {len(items)}"
      )
print(f"Missing covers: {missing_covers}")
print(f"Missing synopses: {missing_synopsis}")
```
