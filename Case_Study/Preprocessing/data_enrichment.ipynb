```python
import requests

from dotenv import load_dotenv
from pathlib import Path
env_path =  'config.env'
load_dotenv(dotenv_path=env_path)

CLIENT_ID = os.getenv("CLIENT_ID")

headers = {
    "X-MAL-CLIENT-ID": CLIENT_ID
}

def get_anime_info(anime_id: int):
    """
    Returns title, synopsis, and cover image URL for a MyAnimeList anime ID.
    """

    response = requests.get(
        f"https://api.myanimelist.net/v2/anime/{anime_id}",
        headers=headers,
        params={
            "fields": "title,synopsis,main_picture"
        }
    )
    
    print("Status:", response.status_code)
    print("Response:", response.text[:500])



    response.raise_for_status()

    data = response.json()

    return {
        "title": data["title"],
        "synopsis": data["synopsis"],
        "cover_art": data["main_picture"]["large"]
        if "main_picture" in data and "large" in data["main_picture"]
        else data.get("main_picture", {}).get("medium")
    }

```

```python

# Example: Frieren
anime = get_anime_info(2104)

print("Title:", anime["title"])
print("Cover:", anime["cover_art"])
print("Synopsis:", anime["synopsis"])
```
